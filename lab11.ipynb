```python
# Recurssive function
def fact(n):
    if n < 0 or n == 1:

      return 1
    elif n>1:
     return n * fact(n - 1)
    else:
      print("FACTORIAL OF ONLY POSITIVE INTEGER IS ALLOWED")


num = int(input("Enter a number :"))

print("Factorial:", fact(num))
```
